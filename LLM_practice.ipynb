```python
import time

def ask_ai(question, tries=3):
    for attempt in range(tries):
        try:
            response = client.models.generate_content(
                model="gemini-flash-lite-latest",
                contents=question,
            )
            return response.text
        except errors.APIError as e:
            if e.code in (429, 500, 503):
                print(f"Busy or limit reached (error {e.code}), waiting 10 seconds...")
                time.sleep(10)
            else:
                print("Error that won't fix itself:", e.code, e.message)
                return None
    return None

print(ask_ai("Write a one-line listing for a studio in JVC"))
```

```python
LISTING_TEMPLATE = """You are a professional Dubai real estate copywriter.

Write a listing for this property:
- Type: {beds}-bed apartment
- Area: {location}
- Size: {size} sqft
- Price: AED {price:,}
- Features: {features}

Rules:
- Exactly 3 sentences
- Mention the price
- Do not invent features that are not listed"""

def write_listing(beds, location, size, price, features):
    prompt = LISTING_TEMPLATE.format(beds=beds, location=location, size=size, price=price, features=features)
    return ask_ai(prompt)

print(write_listing(2, "Dubai Marina", 1400, 1200000, "sea view, gym, pool"))
print("==========")
print(write_listing(1, "JVC", 800, 650000, "balcony, parking"))
```

```python
from pydantic import BaseModel

class Property(BaseModel):
    location: str
    beds: int
    price: int

response = client.models.generate_content(
    model="gemini-flash-lite-latest",
    contents="""Extract the property details. Price must be a full number in AED.
Text: "Lovely 2 bedroom flat in Dubai Marina, asking 1.2 million dirhams." """,
    config=types.GenerateContentConfig(
        response_mime_type="application/json",
        response_schema=Property,
    ),
)

prop = response.parsed
print(prop)
print("Price:", prop.price)
print("Price + 10%:", prop.price * 1.1)
```

```python
class Lease(BaseModel):
    tenant_name: str
    landlord_name: str
    property_address: str
    annual_rent: int
    number_of_cheques: int
    start_date: str
    end_date: str
    security_deposit: int | None = None

lease_text = """
This tenancy agreement is made between Mr. Ahmed Al Mansoori (Landlord)
and Ms. Priya Sharma (Tenant) for Apartment 1204, Marina Heights, Dubai Marina.
The annual rent is AED 95,000 payable in 4 cheques. The tenancy starts on
1 January 2027 and ends on 31 December 2027.
"""

response = client.models.generate_content(
    model="gemini-flash-lite-latest",
    contents=f"""Extract the lease details. Dates must be in YYYY-MM-DD format.
If a detail is not in the text, leave it empty.

Lease:
{lease_text}""",
    config=types.GenerateContentConfig(
        response_mime_type="application/json",
        response_schema=Lease,
    ),
)

lease = response.parsed
print(lease)
```

```python
lease_texts = [
    """Landlord Mr. Ahmed Al Mansoori rents Apartment 1204, Marina Heights, Dubai Marina
    to Ms. Priya Sharma. Rent AED 95,000 per year in 4 cheques, from 1 Jan 2027 to 31 Dec 2027.""",

    """Tenant John Smith will rent Villa 7, Arabian Ranches from landlord Ms. Fatima Khan.
    Yearly rent AED 210,000, 2 cheques. Start 15 March 2027, end 14 March 2028.
    Security deposit AED 10,500.""",

    """Studio 305, JVC Residence is let by Omar Hassan to Li Wei for AED 48,000 annually,
    paid in 12 cheques. Lease period: 1 June 2027 to 31 May 2028.""",
]

rows = []
for text in lease_texts:
    response = client.models.generate_content(
        model="gemini-flash-lite-latest",
        contents=f"""Extract the lease details. Dates must be in YYYY-MM-DD format.
If a detail is not in the text, leave it empty.

Lease:
{text}""",
        config=types.GenerateContentConfig(
            response_mime_type="application/json",
            response_schema=Lease,
        ),
    )
    rows.append(response.parsed.model_dump())

leases_df = pd.DataFrame(rows)
leases_df
```
